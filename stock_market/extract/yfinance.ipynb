```python
import re

def camel_to_snake(name):
    # Insert an underscore between a lowercase/number and an uppercase letter
    name = re.sub(r'(.)([A-Z][a-z]+)', r'\1_\2', name)
    # Handle acronyms and standard camelCase transitions
    return re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', name).lower()

# Examples:
print(camel_to_snake('abcDef'))        # Output: abc_def
print(camel_to_snake('CamelCase'))     # Output: camel_case
print(camel_to_snake('HTTPResponse'))  # Output: http_response
```

```python
target_tickers = yf.Tickers(tickers_code)
metadata_list = []

# Loop through the dictionary of Ticker objects
for symbol, ticker in target_tickers.tickers.items():
    info = ticker.info
    metadata_list.append({
        "ticker": symbol,
        **info
    })
    break

# Convert to a DataFrame for easy viewing
df = pd.DataFrame(metadata_list)
display(df)
```

# Actions

```python
actions_all = pd.DataFrame()
for symbol, ticker in yfinance.tickers.items():
    actions = pd.DataFrame(ticker.actions)
    actions["symbol"] = symbol

    actions_all = pd.concat([actions_all, actions])
    
```

# Capital Gain

```python
shares_full = pd.DataFrame()
for symbol, ticker in yfinance.tickers.items():
    sf = pd.DataFrame(ticker.get_shares_full())
    sf["symbol"] = symbol
    sf.rename(columns={0: "shares_outstanding"}, inplace=True)

    shares_full = pd.concat([shares_full, sf])
    
```
